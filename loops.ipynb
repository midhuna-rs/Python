### nested loop

```python
for i in range(1,5):
    for j in range(2,6):
        print(i,j)
    print()
```

```python
for i in range(5):
    for j in range(8):
        print(i,j,end="   ")    # end=" " prevents print(i, j) from going to the next line. Instead, it adds spaces after each pair
    print()
```

```python
for i in range(5):
    for j in range(i):
        print(i,end=" ")
    print()
```

```python
l=[12,34,56,99,31]
even=0
odd=0
i=0
while i<len(l):
    if l[i]%2==0:
       even+=1
    else:
        odd+=1
    i+=1
print("number of even numbers:",even)
print("number of odd numbers:",odd)
```

### loop control statements

break-exits loop immediately,skipping everything after it

continue-skips that specific iteration and move to next

pass-does nothing , act as a placeholder

```python
for n in range(1,10):
    if n==3:
        break
    print(n)
```

```python
for n in range(10):
    if n==3:
        continue
    print(n)
```

```python
for p in range(10):
    if p==3:
        pass
    print(p)
```

Write a program that asks the user for a number and calculates the sum of all its digits using a while loop.

```python
num=int(input("enter a number:"))
sum=0
while num>0:
    n=num%10
    num=num//10
    sum+=n
print("sum:",sum)
```

Write a program to reverse a number using a while loop.

```python
number=98547
reverse=0
while number>0:
    n=number%10
    number=number//10
    reverse=reverse*10+n
print(reverse)
    
```
