## File Handling

### write

```python
with open("w_demo.txt","w")as f:
    f.write("First write using 'w' mode\n")

with open("w_demo.txt","r")as f:
    print(f.read())
```

## append
Append mode ("a") opens a file so new data is written at the end, preserving its existing contents.
- If the file doesn’t exist, Python creates it.

```python
with open("a_demo.txt","a")as f:
    f.write("First entry\n")
    f.write("second entry\n")
with open("a_demo.txt","r")as f:
    print(f.read())
```

```python
for i in range(2,6):
    with open("f_demo.txt","w")as f:
        f.write(f"Entry {i}\n")         #each time rewrites i

with open("f_demo.txt","r")as f:
    print(f.read())
```

```python
for i in range(2,6):
    with open("f_demo.txt","a")as f:          #a
        f.write(f"Entry {i}\n")         

with open("f_demo.txt","r")as f:
    print(f.read())
```

## exclusive mode x
Create a new file; error if it already exists

```python
with open("f_demo.txt","x")as f:
    f.write("created using f mode\n")
print("file created successfully")
```

```python
import os
if os.path.exists("f_demo.txt"):
    os.remove("f_demo.txt")

with open("f_demo.txt","x")as f:
    f.write("created using f mode\n")
print("file created successfully")
```

FileExistsError

```python
try:
    with open("f_demo.txt","x")as f:
     f.write("created using f mode\n")
    print("file created successfully")

except FileExistsError:
   print("File already exists")

```

the modes as what happens when the file opens and where new text goes:
- "w+": Start with an empty file. You can read and write.
- "r+": Keep the existing file and its contents. You can read and write; the file must already exist.
- "a+": Keep the existing contents, and new text always goes at the end. Creates the file if it doesn’t exist.
The main danger is "w+": opening it erases the old contents. 
- All three modes allow reading, but where reading starts depends on the file position;
- use f.seek(0) to go to the beginning.

## a+ mode append and read

```python
with open("a_demo.txt","w")as f:
    f.write("existing content\n")

with open("a_demo.txt","a+")as f:
    f.write("appended content\n")
    f.seek(0)
    print(f.read())
```

```python
lin=["apple\n","banana\n","mango\n"]
print(lin)

with open("wl_demo.txt","w+")as f:
    f.writelines(lin)
    f.seek(0)
    print(f.read())
```

```python
with open("wl_demo.txt","r")as f:
    lin=f.readlines()
del lin[2]
with open("wl_demo.txt","w")as f:
    f.writelines(lin)

with open("wl_demo.txt","r")as f:
    print(f.read())
```

```python
import os
for filename in ["w_demo.txt","wl_demo.txt","a_demo.txt"]:
    if os.path.exists(filename):
        os.remove(filename)
print("file cleaned up")
```
