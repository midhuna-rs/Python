## Reshaping Arrays

reshape() changes the shape of an array without changing its elements.

```python
arr=np.arange(12)
print(arr)

reshaped=arr.reshape(3,4)
print(reshaped)

reshaped2=arr.reshape(6,2)
print(reshaped2)

reshaped3=arr.reshape(2,2,3)
print(reshaped3)
```

## Transpose
Transpose means turning rows into columns and columns into rows.

```python
arr=np.array([[1,2,3],[4,7,8]])
print(arr)

print(arr.T)
print(np.transpose(arr))


print(arr)
print(arr.sum(axis=0))    #column wise
print(arr.sum(axis=1))    #row wise
```

## Concatenate

np.concatenate() is used to join two or more arrays together.

```python
a=np.array([[1,2],[5,6]])
b=np.array([[8,9],[5,7]])

print(a)
print(b)

print(np.concatenate(a))
print(np.concatenate(b))

print(np.concatenate((a,b),axis=0))
print(np.concatenate((a,b),axis=1))
```

## Mean , Median , Standard Deviation
- Mean means adding all the numbers and dividing by how many numbers there are.
- Median is the middle number after sorting the numbers in ascending order.
- Standard deviation tells us how far the values tend to spread around their mean.

```python
arr=np.array([1,2,3,4,5,6])
print("mean:",np.mean(arr))
print("median:",np.median(arr))
print("std:",round(np.std(arr),1))
```

### Student score analysis

```python
scores=np.array([78,45,89,92,34,67,55,90,23,88])

print("Average:",np.mean(scores))
print("Highest Score:",np.max(scores))
print("Lowest Score:",np.min(scores))
print("Pass(>=40):",scores[scores>=40])
print("Number of pass:",np.sum(scores>=40))
print("Number of fails:",np.sum(scores<40))
```

### Monthly sales Analysis

```python
sales=np.array([12000,15500,9800,17200,14300,19800,16700,15200,18100,20500,21300,23400])
months=np.array(["Jan","Feb","Mar","Apr","May","June","July","Aug","Sep","Oct","Nov",'Dec'])

print("Total sales:",np.sum(sales))
print("Average Monthly sales:",np.mean(sales))
best_mont=months[np.argmax(sales)]
print("Best Sales Month:",best_mont, "-->" ,np.max(sales))
```

### Stock Price Daily Returns

```python
prices=np.array([100,102,101,105,110,108,115,120])
daily_returns=np.diff(prices)/prices[:-1]*100
print(np.diff(prices))
print("Daily Returns(%):",np.round(daily_returns,2))

print("Average daily return:",np.round(np.mean(daily_returns),2))
print("best day return:",np.round(np.max(daily_returns),2))
print("Worst day return:",np.round(np.min(daily_returns),2))
```
