```python
def show():
    message="hello"     #local variable
    print(message)

show()
#print(message)          #nameeerror
```

### global variable
A global variable is created outside any function and can be accessed throughout the module.


```python
greet="hello"
def greetings():
 print(greet)

greetings()

print(greet)
```

To reassign it inside a function, use the global keyword.

```python
score=10
print("Global variable:",score)
def show():
    global score
    score=20
    print("Global variable after reassigning in function:",score)

show()
print("Global variable outside the function:",score)   #After a function reassigns a global variable using global, the new value is visible wherever that global variable is used in the module.
```

LEGB is the order Python searches for a variable: 
- Local (current function), 
- Enclosing (a containing function), 
- Global (outside functions), 
- then Built-in (names like print and len). 
- It uses the first matching name it finds.

```python
def outer():
    message = "hello"

    def inner():
        message = "hi"
        print("inner:", message)

    inner()
    print("outer:", message)

outer()
```

The new line is nonlocal message. It tells inner() to change the message from outer() instead of making its own separate local message

```python
def outer():
    message = "hello"

    def inner():
        nonlocal message
        message = "hi"
        print("inner:", message)

    inner()
    print("outer:", message)

outer()
```

For changing a variable:
- Outside all functions? Use global.
- In a containing function? Use nonlocal.
- Otherwise, Python creates or uses a local variable.


- global lets a function reassign a variable defined outside all functions.
- nonlocal lets a nested function reassign a variable in its nearest containing function.
local and enclosing describe scopes; they aren’t keywords

Local variables are created by assignment inside a function. You don’t write a local keyword.

```python
def outer():
    x = "before"
    print(x)

    def inner():
        nonlocal x
        x = "after"

    inner()
    print(x)

outer()  # after
```

## Recursive Function

the pattern: a recursive function solves a small case, then combines that result with the current input.
- A recursive function is a function that calls itself with a smaller or simpler input. It needs a base case to stop and a recursive step that moves toward that case.

```python
#factorial

def fact(n):
    if n==1:
        return 1
    else:
        return n*fact(n-1)
fact(3)
```

```python
#power

def power(base,exp):
    if exp==0:
        return 1
    else:
        return base*power(base,exp-1)
power(3,2)
```

```python
def countdown(count):
    if count==0:
        return 0
    else:
       print(count)
       return countdown(count-1)
        
print(countdown(3))
```

## lambda function
A lambda function is a small, anonymous function written as a single expression. It takes input values, calculates an expression, and automatically returns the result.

lambda inputs: expression

```python
sum = lambda a,b:a+b
sum(9,4)
```

```python
greet= lambda a:print("hello",a)
greet("Anu")
```

conditional lambda
- value_if_true if condition else value_if_false


```python
oddeven=lambda a: "even" if a%2==0 else "odd"
oddeven(3)
```

## with map() apply lambda function  to every item in a list

list(map(lambda item: expression, iterable))

```python
l=[1,2,3,4]
print(list(map(lambda x:x+2,l)))
```

## filter() is for keeping only the items that meet a condition.

```python
li=[1,2,3,4,5,8,7,9]
print(list(filter(lambda x:x%2==0,li)))
```
