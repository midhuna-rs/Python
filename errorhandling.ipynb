```python
def process(value):
    try:
        number=int(value)
        result=100/number
        print("result:",result)
    except ValueError :
        print(f"'{value}' is not a valid number")
    except ZeroDivisionError:
        print("cannot divide by zero")

process(10)
process("abc")
process(0)
process("5.5")
```

## else clause

```python
def positive(num):
    try:
         assert num>0
    except AssertionError:
         print(num,"is not positive")
    else:
         print(num,"is positive")

positive(9)
positive(-1)
```

```python
def divide_with_else(a,b):
    try:
        result=a/b
    except ZeroDivisionError:
        print("Error:division by zero")
    else:
        print("Result:",result)

divide_with_else(12,24)
divide_with_else(2,0)
```

## finally clause
block that always runs

```python
def div(a,b):
    try:
        result=a/b
    except ZeroDivisionError:
        print("division by zero")
    else:
        print("division successful result:",result)
    finally:
        print("execution completed \n")

div(2,4)
div(12,0)
```
